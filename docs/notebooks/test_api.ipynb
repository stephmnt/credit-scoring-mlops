```python
import requests
from urllib.parse import urlparse

"""SPACE_URL = "https://huggingface.co/spaces/stephmnt/credit-scoring-mlops"
# Alternative (direct Space URL): "https://stephmnt-credit-scoring-mlops.hf.space"


def resolve_space_base(url: str) -> str:
    url = url.rstrip("/")
    if "hf.space" in url:
        return url
    parsed = urlparse(url)
    parts = parsed.path.strip("/").split("/")
    if len(parts) >= 3 and parts[0] == "spaces":
        owner, space = parts[1], parts[2]
        return f"https://{owner}-{space}.hf.space"
    raise ValueError(f"Unsupported Space URL: {url}")"""


BASE_URL = "https://stephmnt-credit-scoring-mlops.hf.space"
API_BASE = f"{BASE_URL}/api"

r = requests.get(
    f"{API_BASE}/health",
    headers={"Accept": "application/json"},
)
print("API_BASE:", API_BASE)
print(r.status_code)
print(r.json())
```

```python
s = requests.get(
    f"{API_BASE}/features",
    headers={"Accept": "application/json"},
)
print(s.json())
```

```python
url = f"{API_BASE}/predict"
params = {"threshold": 0.5}

payload = {
    "data": {
        "SK_ID_CURR": 100002,
        "EXT_SOURCE_2": 0.61,
        "EXT_SOURCE_3": 0.75,
        "AMT_ANNUITY": 24700.5,
        "EXT_SOURCE_1": 0.45,
        "CODE_GENDER": "F",
        "DAYS_EMPLOYED": -207,
        "AMT_CREDIT": 406597.5,
        "AMT_GOODS_PRICE": 351000.0,
        "DAYS_BIRTH": -9461,
        "FLAG_OWN_CAR": "N",
        "DAYS_LAST_PHONE_CHANGE": -5,
        "REGION_RATING_CLIENT_W_CITY": 2,
    }
}

r = requests.post(
    url,
    params=params,              # ?threshold=0.5
    json=payload,               # équivalent -d + Content-Type
    headers={"Accept": "application/json"}
)

print(r.status_code)
print(r.json())
```
